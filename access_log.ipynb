```python

import random
import time
import os
from datetime import datetime, timezone, timedelta


browser = [
    "Chrome/130.0",
    "Firefox/131.0",
    "Safari/17.0",
    "Edge/130.0",
    "PostmanRuntime/7.36.0",
    "curl/8.5.0",
    "python-requests/2.31.0"
]

percorsi = [
    "/", "/login", "/dashboard", "/admin",
    "/wp-admin/", "/api/login", "/api/users",
    "/css/style.css", "/images/logo.png",
    "/nonexistent"
]

metodi = ["GET", "POST"]

codici_http = [200, 200, 200, 302, 403, 404, 500]

utenti = ["mario", "anna", "luigi", "giulia", "-"]

referer = [
    '"-"',
    '"https://example.com/"',
    '"https://example.com/login"',
    '"https://example.com/home"'
]


cartella = "logs"
os.makedirs(cartella, exist_ok=True)

percorso_file = os.path.join(cartella, "access_log.txt")


try:
    with open(percorso_file, "a", encoding="utf-8") as file:

        while True:

            ip = random.choice([
                f"192.168.1.{random.randint(1, 254)}",
                f"10.0.0.{random.randint(1, 254)}",
                f"172.16.0.{random.randint(1, 254)}",
                ".".join(str(random.randint(1, 255)) for _ in range(4))
            ])

            identita = "-"
            utente = random.choice(utenti)

            data = datetime.now().astimezone().strftime(
                "%d/%b/%Y:%H:%M:%S %z"
            )

            richiesta = (
                f'"{random.choice(metodi)} '
                f'{random.choice(percorsi)} HTTP/1.1"'
            )

            status = random.choice(codici_http)
            bytes_risposta = random.randint(100, 10000)

            provenienza = random.choice(referer)
            user_agent = f'"Mozilla/5.0 ({random.choice(browser)})"'

            riga = (
                f'{ip} {identita} {utente} [{data}] '
                f'{richiesta} {status} {bytes_risposta} '
                f'{provenienza} {user_agent}\n'
            )

            file.write(riga)
            file.flush()

            print(riga.strip())

            time.sleep(1)

except KeyboardInterrupt:
    print("\nGenerazione dei log interrotta.")

print(f"File salvato in: {percorso_file}")
```
